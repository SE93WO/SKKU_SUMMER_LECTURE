## 0. 환경 준비

- `pandas` : 표 형태 데이터 처리
- `TimeSeriesDataFrame` : 시계열 데이터를 담는 형식
- `TimeSeriesPredictor` : 시계열 모델 학습·예측 (다음 실습에서 사용)

````python
import pandas as pd
from autogluon.timeseries import TimeSeriesDataFrame, TimeSeriesPredictor

print("준비 완료")
````

---
## 1. 데이터 불러오기

- 앞선 실습과 동일한 따릉이 파일 사용
- `encoding="latin-1"` : 컬럼명의 특수문자(°C) 처리
- `format="%d/%m/%Y"` : 날짜가 '일/월/년' 순서임을 지정

````python
df = pd.read_csv("../SeoulBikeData.csv", encoding="latin-1")
df["Date"] = pd.to_datetime(df["Date"], format="%d/%m/%Y")

print("원본 크기:", df.shape)
print("컬럼 목록:")
for col in df.columns:
    print("  -", col)
````

#### 확인 사항
- 원본은 **시간별** 기록 (8,760행 = 365일 × 24시간)
- 14개 열: 날짜·시간 + 대여량 + 기상 변수 8종 + 범주형 3종
- 본 실습은 **일별** 예측을 수행 → 하루 단위로 집계 필요

---
## 2. 일별 집계 — 변수마다 다른 집계 방식

- 시간별 기록을 하루 단위로 합침
- **변수마다 적절한 집계 방식이 다름** → 세 갈래로 구분됨

| 집계 방식 | 해당 변수 | 이유 |
|---|---|---|
| **합계 (sum)** | 대여량, 강수량, 적설량, 일사량 | 하루 동안 누적된 총량이 의미를 가짐 |
| **평균 (mean)** | 기온, 습도, 풍속, 가시거리, 이슬점 | 하루 동안의 대표 상태 (더하면 무의미) |
| **첫 값 (first)** | 계절, 공휴일, 운영일 | 하루 내내 동일한 값 |

- 예: 하루 대여량을 **평균** 내면 의미 없음 / 하루 기온을 **합계** 내면 의미 없음
- 어떤 방식이 적절한지는 **데이터의 의미를 아는 연구자만 판단 가능**
- `agg({...})` : 열마다 다른 집계 방식 지정

````python
agg_rule = {
    # 합계: 하루 동안 누적된 총량
    "Rented Bike Count": "sum",
    "Rainfall(mm)": "sum",
    "Snowfall (cm)": "sum",
    "Solar Radiation (MJ/m2)": "sum",
    # 평균: 하루 동안의 대표 상태
    "Temperature(°C)": "mean",
    "Humidity(%)": "mean",
    "Wind speed (m/s)": "mean",
    "Visibility (10m)": "mean",
    "Dew point temperature(°C)": "mean",
    # 첫 값: 하루 내내 동일한 범주
    "Seasons": "first",
    "Holiday": "first",
    "Functioning Day": "first",
}

daily = df.groupby("Date").agg(agg_rule).round(2).reset_index()

print("일별 집계 크기:", daily.shape)
daily.head()
````

#### 확인 사항
- 8,760행 → **365행**으로 압축됨 (열은 14개 → 13개, `Hour`는 집계로 사라짐)
- 2017년 12월 1일: 대여량 9,539대(합계), 기온 -2.45°C(평균), 계절 Winter(첫 값)
- 각 변수가 의미에 맞게 집계됨

> **주의:** `Functioning Day`는 하루 안에서 값이 바뀌는 날이 존재함 (일부 시간만 운영 중단)
> - `first`는 그날의 첫 시간 값만 취함 → 완전히 정확하지는 않음
> - 이러한 판단 역시 연구자의 몫임

---
## 3. 시계열 형식으로 열 정리

- AutoGluon 시계열의 **기본 3열**

| 열 | 의미 | 따릉이 |
|---|---|---|
| `item_id` | 어느 시계열인가 (구분용 이름표) | 따릉이는 하나 → 모두 동일 |
| `timestamp` | 언제인가 (시점) | 날짜 |
| `target` | 얼마인가 (예측할 값) | 일별 대여량 |

- **추가 변수**(기상·범주형)는 열로 그대로 유지함
- `rename(columns=...)` : 열 이름 변경 (특수문자 제거)
- `item_id` 열 추가 : 시계열이 하나이므로 모두 `"seoul_bike"`로 채움

````python
ts_df = daily.rename(columns={
    "Date": "timestamp",
    "Rented Bike Count": "target",
    "Temperature(°C)": "temperature",
    "Humidity(%)": "humidity",
    "Wind speed (m/s)": "wind_speed",
    "Visibility (10m)": "visibility",
    "Dew point temperature(°C)": "dew_point",
    "Solar Radiation (MJ/m2)": "solar_radiation",
    "Rainfall(mm)": "rainfall",
    "Snowfall (cm)": "snowfall",
    "Seasons": "seasons",
    "Holiday": "holiday",
    "Functioning Day": "functioning_day",
})
ts_df["item_id"] = "seoul_bike"

# 기본 3열을 앞으로 정렬
cols = ["item_id", "timestamp", "target"] + [
    c for c in ts_df.columns if c not in ["item_id", "timestamp", "target"]
]
ts_df = ts_df[cols]

print("크기:", ts_df.shape)
ts_df.head()
````

#### 확인 사항
- 기본 3열(`item_id`·`timestamp`·`target`) + 추가 변수 10열
- 추가 변수는 **버리지 않고 모두 함께 넣음** → 예측 정확도 향상에 기여 가능

> **참고: 여러 시계열을 다루는 경우**
> - `item_id`로 구분함 (예: 매장별 매출 → `매장A`, `매장B` …)
> - 각 시계열을 개별적으로 예측함

---
## 4. TimeSeriesDataFrame으로 변환

- AutoGluon 전용 형식으로 변환
- `from_data_frame(...)` : 일반 표를 시계열 형식으로 변환
- `id_column` : 시계열 구분 열 지정
- `timestamp_column` : 시점 열 지정
- 나머지 열(`target` + 추가 변수)은 **그대로 유지됨**

````python
ts_data = TimeSeriesDataFrame.from_data_frame(
    ts_df,
    id_column="item_id",
    timestamp_column="timestamp"
)

print("크기:", ts_data.shape)
ts_data.head()
````

#### 확인 사항
- `item_id`와 `timestamp`가 **인덱스**로 이동함 (왼쪽 축)
- `target`과 추가 변수 10개는 열로 남음
- 이것이 AutoGluon이 요구하는 시계열 형식임

> **어느 열이 예측 대상인가?**
> - 모델 학습 시 `TimeSeriesPredictor(target="target")`로 지정함
> - 나머지 열은 **자동으로 보조 변수로 인식됨**
> - 미래 값을 미리 아는 변수(공휴일 등)는 `known_covariates_names`로 별도 지정 가능

---
## 5. 시간 간격 확인

- AutoGluon이 타임스탬프를 보고 시간 간격을 **자동 추론**함
- `.freq` : 추론된 간격 확인
- `"D"` = 일별(Daily), `"h"` = 시간별(hourly), `"ME"` = 월별

````python
print("시간 간격:", ts_data.freq)
print("시계열 개수:", len(ts_data.item_ids))
print("시점 개수:", ts_data.num_timesteps_per_item().values)
````

#### 확인 사항
- 시간 간격 = `D` → 일별 데이터로 자동 인식됨
- 시계열 개수 = 1 → 따릉이 하나
- 시점 개수 = 365 → 1년치

> **주의:** 자동 추론은 속도를 위해 **시계열의 일부만 확인**하므로 간혹 부정확할 수 있음
> - 정확한 확인이 필요하면 `infer_frequency()` 사용
> - 간격이 불규칙하면 `.freq`는 `None`을 반환함
> - 이 경우 `TimeSeriesPredictor(freq="D")`로 직접 지정함

---
## 6. 시간 간격 변경 (집계)

- `convert_frequency(freq=...)` : 시간 간격을 다른 단위로 변경함
- 예: 시간별 → 일별, 일별 → 주별, 분기별 → 연별
- `agg_numeric` : 값을 합치는 방식 (`"sum"`, `"mean"`, `"max"`, `"min"`, `"median"` 등)
- **기본값은 `"mean"`(평균)**

````python
# 일별 → 주별로 집계
ts_weekly = ts_data.convert_frequency(freq="W", agg_numeric="sum")

print("일별 시점 개수:", ts_data.num_timesteps_per_item().values)
print("주별 시점 개수:", ts_weekly.num_timesteps_per_item().values)
ts_weekly.head()
````

#### 결과 해석
- 365개(일별) → 약 52개(주별)로 집계됨
- 명령 한 줄로 시간 단위 변경 완료
- pandas의 `groupby().resample()`보다 빠름 (여러 CPU 코어로 병렬 처리)

> **주의: 모든 수치 열에 같은 방식이 적용됨**
> - 위 결과에서 `temperature`·`humidity` 등도 **합산**됨 → 주간 기온 합계는 의미 없음
> - 변수마다 다른 집계가 필요하면 pandas의 `groupby().agg({...})` 사용 (2단계 방식)
> - 즉, **변수가 하나면 `convert_frequency`, 여러 개면 pandas**가 적합함

---
## 7. 빠진 시점의 자동 처리

- 실제 데이터에는 특정 날짜가 통째로 누락되는 경우가 존재함
- **같은 `convert_frequency` 명령**이 빠진 시점 채우기도 수행함
- 확인을 위해 임의로 날짜 3개를 제거한 뒤 처리 결과를 관찰함

````python
# 확인용: 12월 3~5일을 임의로 제거
missing_df = ts_df.drop(index=[2, 3, 4])

ts_missing = TimeSeriesDataFrame.from_data_frame(
    missing_df,
    id_column="item_id",
    timestamp_column="timestamp"
)
print("제거 후 시점 개수:", ts_missing.num_timesteps_per_item().values)
ts_missing[["target", "temperature"]].head()
````

````python
# 빠진 시점 채우기
ts_filled = ts_missing.convert_frequency(freq="D")

print("채운 후 시점 개수:", ts_filled.num_timesteps_per_item().values)
ts_filled[["target", "temperature"]].head(6)
````

#### 결과 해석
- 제거했던 12월 3~5일이 **행으로 다시 삽입됨**
- 단, 값은 `NaN`(비어 있음) 상태 → **임의로 값을 지어내지 않음**
- 대부분의 시계열 모델은 `NaN`을 자체적으로 처리하므로 **수동 처리 불필요**

> **연구자의 개입이 필요한 경우:** 결측의 **의미**는 연구자만 판단 가능함
> - "측정하지 못함" → 그대로 두면 모델이 처리함
> - "수요가 0이었음" → 0으로 채우는 것이 적절함
> ```python
> ts_filled.fill_missing_values(method="constant", value=0)
> ```

## 정리

| 단계 | 명령 | 내용 |
|---|---|---|
| 일별 집계 | `groupby("Date").agg({...})` | 시간별 → 일별 · 변수마다 다른 집계 방식 |
| 열 정리 | `rename`, `item_id` 추가 | 기본 3열 + 추가 변수 10개 |
| 형식 변환 | `TimeSeriesDataFrame.from_data_frame()` | AutoGluon 전용 형식으로 |
| 간격 확인 | `.freq` | 일별(`D`)로 자동 인식 |
| 간격 변경 | `convert_frequency(freq="W", agg_numeric="sum")` | 일별 → 주별 집계 |
| 빠진 시점 | `convert_frequency(freq="D")` | 자동으로 행 삽입 (값은 NaN) |

**요약**
- 시계열의 기본은 **세 개의 열**(item_id · timestamp · target)
- 기상 변수 등 **추가 변수는 열로 모두 함께 넣음** → 버릴 필요 없음
- 시간 간격 추론·간격 변경·빠진 시점 채우기는 **AutoGluon이 수행**

**연구자의 몫**
- 시간 단위 결정 (시간별? 일별? 월별?)
- **집계 방식 결정** (합계? 평균? 첫 값?) — 변수의 의미에 따라 다름
- 결측의 의미 판단 (측정 실패인가, 0인가)
